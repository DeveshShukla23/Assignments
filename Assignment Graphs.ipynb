```python
df = px.data.gapminder().query("year==2007").query("continent=='Europe'")
df.loc[df['pop'] < 2.e6, 'contry'] = 'Other countries'
fig = px.pie(df, values = 'pop', names = 'country', title = 'Population of European continent', width = 600, height = 600)
fig.show()
```

```python
df = px.data.gapminder().query("year==2007").query("continent=='Asia'")
df.loc[df['pop'] < 2.e6, 'contry'] = 'Other countries'
fig = px.pie(df, values = 'pop', names = 'country', title = 'Population of European continent', width = 700, height = 700)
fig.show()
```

```python
df = px.data.gapminder().query("year==2007").query("continent=='Europe'")
fig = px.sunburst(df, path = ['continent', 'country'], values = 'pop', color = 'lifeExp'
                 , hover_data = ['iso_alpha'])
fig.show()
```
